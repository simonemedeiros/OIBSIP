```python
# Google Play Store Analysis

# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from textblob import TextBlob

from pathlib import Path

# Project paths

PROJECT_DIR = Path.cwd().parent

RAW_DATA_DIR = PROJECT_DIR / "data" / "raw"

APPS_FILE = RAW_DATA_DIR / "googleplaystore.csv"
REVIEWS_FILE = RAW_DATA_DIR / "googleplaystore_user_reviews.csv"

# Load Datasets

apps = pd.read_csv(APPS_FILE)
reviews = pd.read_csv(REVIEWS_FILE)

# Dataset Overview

print(f"Apps dataset: {apps.shape}")

print(f"Reviews dataset: {reviews.shape}")

apps.head()
```

```python
# 2. Data Quality Assessment

# 2.1 Dataset structure
apps.info()

reviews.info()

# 2.2 Missing values
print("Missing values - Apps")
apps.isnull().sum().sort_values(ascending = False)

print("Missing values - Reviews")
reviews.isnull().sum().sort_values(ascending = False)

# 2.3 Duplicate rows
print(f"Duplicate rows in apps: {apps.duplicated().sum()}")
print(f"Duplicate rows in reviews: {reviews.duplicated().sum()}")

# 2.4 Unique values and basic inspection
apps.nunique().sort_values()
reviews.nunique().sort_values()

# 2.5 Numeric and categorical columns
apps.describe(include = "all").T
reviews.describe(include = "all").T

# 2.6 Inspect problematic columns
apps[["Reviews", "Installs", "Price", "Size", "Rating"]].head(10)
apps["Installs"].unique()[:20]
apps["Price"].unique()[:20]
apps["Size"].unique()[:20]

# 2.7 Review dataset
reviews.head()
reviews["Sentiment"].value_counts(dropna=False)
```
