```python
from pathlib import Path

path = Path("/content/Time-Series-Library/run.py")
text = path.read_text()

old_seed = """    fix_seed = 2021
    random.seed(fix_seed)
    torch.manual_seed(fix_seed)
    np.random.seed(fix_seed)
"""

if old_seed not in text:
    raise RuntimeError("Bloco original da seed não encontrado.")

text = text.replace(old_seed, "")

marker = "    args = parser.parse_args()\n"

if marker not in text:
    raise RuntimeError("Linha parser.parse_args() não encontrada.")

new_block = """    args = parser.parse_args()

    # Reproducible experimental seed
    random.seed(args.seed)
    np.random.seed(args.seed)
    torch.manual_seed(args.seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(args.seed)

    print(f"Experimental seed: {args.seed}")
"""

text = text.replace(marker, new_block, 1)
path.write_text(text)

print("run.py modificado com sucesso.")
```

```python
%%writefile /content/Time-Series-Library/run_experiments.py

import subprocess
from pathlib import Path

repo = Path("/content/Time-Series-Library")

# k=1/seed=2021 has already been completed.
experiments = [
    (k, seed)
    for k in [1, 2, 5]
    for seed in [2021, 2022, 2023, 2024, 2025]
    if not (k == 1 and seed == 2021)
]

for number, (k, seed) in enumerate(experiments, start=2):

    model_id = f"weather_96_720_k{k}_seed{seed}"

    result_pattern = list(
        (repo / "results").glob(
            f"long_term_forecast_{model_id}_TimesNet_*"
        )
    )

    # Skip completed experiments.
    if result_pattern:
        metrics_file = result_pattern[0] / "metrics.npy"

        if metrics_file.exists():
            print(
                f"[SKIP] {number}/15 | "
                f"top_k={k} | seed={seed} already completed",
                flush=True,
            )
            continue

    print(
        f"\n{'=' * 70}\n"
        f"EXPERIMENT {number}/15 | top_k={k} | seed={seed}\n"
        f"{'=' * 70}",
        flush=True,
    )

    command = [
        "python", "-u", "run.py",
        "--task_name", "long_term_forecast",
        "--is_training", "1",
        "--root_path", "./dataset/weather/",
        "--data_path", "weather.csv",
        "--model_id", model_id,
        "--model", "TimesNet",
        "--data", "custom",
        "--features", "M",
        "--seq_len", "96",
        "--label_len", "48",
        "--pred_len", "720",
        "--e_layers", "2",
        "--d_layers", "1",
        "--factor", "3",
        "--enc_in", "21",
        "--dec_in", "21",
        "--c_out", "21",
        "--d_model", "32",
        "--d_ff", "32",
        "--top_k", str(k),
        "--des", "Exp",
        "--itr", "1",
        "--train_epochs", "1",
        "--batch_size", "32",
        "--num_workers", "0",
        "--seed", str(seed),
    ]

    subprocess.run(
        command,
        cwd=repo,
        check=True,
    )

print("\nAll requested experiments completed.", flush=True)
```

```python
%cd /content/Time-Series-Library

from pathlib import Path
import numpy as np

results_dir = Path("results")

found = []

for k in [1, 2, 5]:
    for seed in [2021, 2022, 2023, 2024, 2025]:
        model_id = f"weather_96_720_k{k}_seed{seed}"

        folders = list(
            results_dir.glob(
                f"long_term_forecast_{model_id}_TimesNet_*"
            )
        )

        metrics_files = [
            folder / "metrics.npy"
            for folder in folders
            if (folder / "metrics.npy").exists()
        ]

        if metrics_files:
            metrics = np.load(metrics_files[0])

            print(
                f"✅ k={k} seed={seed} | "
                f"metrics={metrics}"
            )

            found.append((k, seed))
        else:
            print(f"❌ k={k} seed={seed}")

print(f"\nTOTAL ENCONTRADO: {len(found)}/15")
```
